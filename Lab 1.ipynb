```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r'C:\users\resea\Downloads\fruits.jpeg')

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')
plt.show()

r,g,b = img[:,:,0], img[:,:,1], img[:,:,2]

plt.figure(figsize=(15,5))
plt.subplot(1,3,1)
plt.imshow(r, cmap='gray')
plt.title("Red Channel Image")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(g, cmap='gray')
plt.title("Green Channel Image")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(b, cmap='gray')
plt.title("Blue Channel Image")
plt.axis('off')
plt.show()

gbr = img
gbr[:,:,0] = g
gbr[:,:,1] = b
gbr[:,:,2] = r
plt.imshow(gbr)
plt.title('Reordered Image (GBR)')
plt.axis('off')
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r'C:\users\resea\Downloads\fruits.jpeg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

invImg = img.copy()

rows, columns, channels = invImg.shape

for i in range(rows):
    for j in range(columns):
        for k in range(channels):
            invImg[i,j,k] = 255 - invImg[i,j,k]


plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(invImg)
plt.title("Inverse Image")
plt.axis('off')

plt.show()
```
